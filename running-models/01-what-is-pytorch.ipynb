## What PyTorch actually is

PyTorch is the **engine that runs most modern AI models**. When you read about a new open model — a language model, an image generator, a speech recognizer — the odds are overwhelming that it was built with PyTorch and ships as PyTorch weights. Learn to drive PyTorch and a huge fraction of the AI world opens up.

Under the hood it's three things:

- **Tensors** — multi-dimensional arrays (think NumPy arrays) that can live on a GPU for speed. This is the data structure everything flows through, and it's the one part you genuinely need to be comfortable with. That's chapter 2.
- **Autograd** — the bookkeeping that lets a model *learn* by automatically working out how to adjust its numbers. It's the clever core of training. Here's the good news: **to run an existing model, you never touch it.**
- **`nn` (neural network building blocks)** — the layers models are assembled from. We'll take a look inside one so models stop feeling like black boxes, but you won't be hand-building architectures.

The useful analogy: PyTorch is like a car. **Autograd and the training math are the engine internals.** You're learning to *drive* — steering, pedals, reading the dashboard — not to rebuild the transmission. Plenty of people drive their whole lives without opening the hood, and they get where they're going.

## What you can safely ignore

Because this trips people up, let me be blunt about what this book deliberately skips:

- **The calculus.** Gradients, derivatives, the chain rule, backpropagation — the machinery of *how* models learn. You can run any pretrained model without it.
- **The training loop.** Epochs, loss functions, optimizers. That's how a model is *made*; we're *using* ones that already exist.
- **The linear algebra proofs.** You need to know a tensor has a *shape*, not how to derive a matrix multiplication.

None of that is forbidden knowledge — it's just a different goal, and it's where the Training Models book goes (understanding and fine-tuning models on your own data). This book keeps both hands on the wheel: load a model, feed it input, read the output, repeat with bigger and more interesting models. If you've been put off PyTorch by a tutorial that opened with partial derivatives, this is the book that doesn't.

## What "running a model" will look like

To show you where this is heading, here's the kind of code this book builds toward — a complete, working sentiment classifier using a pretrained model:


```python
from transformers import pipeline

classify = pipeline("sentiment-analysis")
print(classify("I can't believe how easy this was."))
# [{'label': 'NEGATIVE', 'score': 0.9987}]
```



Five lines, a real model, a real answer — and look at the answer. The sentence was meant as praise, and the model called it negative, with 99.9% confidence. It read "I can't believe" the way it usually appears in its training data: as the start of a complaint. That is not a broken model. It is the first lesson of the book, delivered on line five: a model's output is a *verdict*, produced by arithmetic over what it has seen, and the number beside it measures how sure the arithmetic is, not how right. You don't yet know what `pipeline` is doing or where the model came from — that's exactly what the book unpacks. But notice the shape of it: **load a pretrained thing, hand it an input, read the output.** Almost everything you'll run follows that pattern, and almost everything you'll learn is how to read what comes back.

## Final thoughts

The single most freeing idea in this book is permission to *not* learn the math first. The people shipping AI features into products mostly aren't deriving gradients; they're running capable pretrained models and wiring them into something useful, and that skill is learnable in an afternoon's worth of notebooks. PyTorch is just the engine you're learning to drive — and you can drive it well long before you understand every part under the hood.
